```python
import re
for i in df['Age'].apply(lambda x: re.sub(r'[^0-9]', '', str(x))):
    if len(i) <2:
        print(i)

def extract_age(age):
    age_num = re.findall('[0-9]+', str(age))
    if len(age_num) > 0:
        return age_num[0]
    else:
        return age

df['Age'] = df['Age'].apply(lambda x: extract_age(x))
```

```python
df_age = df[df['Age'] != 'nan years']['Age']
```

```python
df = pd.read_csv("messy_customer_sales_data.csv")
df.dropna(subset=['Customer_ID'], inplace=True)

df['Age'] = df['Age'].replace('nan years', age_median)
df['Age'] = df['Age'].fillna(age_median)

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())
df['Feedback_Score'] = df['Feedback_Score'].fillna(df['Feedback_Score'].mode().iloc[0])

print(df.isnull().sum())
```

```python
for col in ['Gender', 'City', 'Country']:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
df['Last_Purchase_Date'].fillna(df['Last_Purchase_Date'].mode()[0], inplace=True)
```

Handling outliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

cols  = ['Age', 'Purchase_Amount', 'Feedback_Score']
plt.figure(figsize=(15,5))

for i, col in enumerate(cols, 1):
    plt.subplot(1,3,i)
    sns.boxplot(y = df[col], color = 'skyblue', width = 0.3)
    plt.title(f'{col} boxplot', fontsize = 15)
    plt.ylabel('')
    plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
